```python
def R2T_300(R):  # from 6K to 96K
    a0, a1, a2, a3, a4, a5, a6, a7, a8 = -1.60683527e+03, 5.17195992e+03, -5.91765185e+03,  3.80559728e+03, -1.51000611e+03,  3.78589829e+02, -5.85650536e+01, 5.11096728e+00, -1.92704984e-01
    
    if np.isscalar(R):
        R = np.array([R])

    if np.any(R < 294.36) or np.any(R > 326.43):
        print('out of range: results may not be accurate')

    def f(T, R):
        log_T = np.log(T)
        return (a0 + a1 * log_T + a2 * log_T**2 + a3 * log_T**3 +
                a4 * log_T**4 + a5 * log_T**5 + a6 * log_T**6 +
                a7 * log_T**7 + a8 * log_T**8) - R

    def find_temp(R):
        try:
            return brentq(f, 6, 97, args=(R,))
        except ValueError:
            raise ValueError('root finding did not converge within the temperature range 6K to 96K')

    temperatures = np.array([find_temp(R_input) for R_input in R])
    
    if temperatures.size == 1:
        return temperatures[0]
    else:
        return temperatures
```
